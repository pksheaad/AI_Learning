# Understanding the Three Transformer Architectures

Transformers are the foundation of modern Large Language Models (LLMs). There are three main architectures:

**Encoder-only:** Primarily used to understand and represent input text.

**Decoder-only:** Primarily used to generate text, one token at a time.

**Encoder–Decoder:** Used to transform an input sequence into an output sequence, such as translating or summarizing text.

*Typical encoder-only use cases include sentiment analysis, named-entity recognition, text classification, extractive question answering, and creating text embeddings*.

## 1. Encoder-only architecture

An encoder-only model reads the entire input sequence and produces contextual representations of its tokens.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification
import torch
```

```python
model_name = "distilbert-base-uncased-finetuned-sst-2-english"
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSequenceClassification.from_pretrained(model_name)
input_prompt = ["The movie was fantastic.", "The movie was terrible."]
input_tokens = tokenizer(input_prompt, return_tensors = "pt")
input_tokens
with torch.inference_mode():
  output = model(input_tokens['input_ids'])
output.logits.shape
```

```python
probability = torch.softmax(output.logits, dim = -1)
probability
```

```python
sentiment_prediction = torch.argmax(probability, dim = -1)
tokenizer.decode(sentiment_prediction)
```

## Decoder-only architecture
A decoder-only transformer is designed to predict the next token using the tokens that came before it.

Consider:

The capital of France is

The model predicts the next token, perhaps Paris, appends it to the sequence, and continues generating until a stopping condition is reached.

Unlike the encoder, a decoder-only transformer uses causal (masked) self-attention. This prevents a token from seeing future tokens.

Common decoder-only use cases include conversational assistants, code generation, creative writing, question answering, and general-purpose text generation.

```python
from transformers import AutoTokenizer, AutoModelForCausalLM
import torch
```

```python
model_name = "gpt2"
```

```python
tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForCausalLM.from_pretrained(model_name)
```

```python
input_prompt = "The caiptal of India is "
input = tokenizer(input_prompt, return_tensors = "pt")
model.eval()
with torch.inference_mode():
  output = model(**input)
output

probs = torch.softmax(output.logits, dim = -1)
probs

prediction = torch.argmax(probs, dim = -1)
prediction
```

## Encoder–Decoder architecture

An encoder–decoder model combines both architectures. The encoder processes the input sequence, and the decoder generates a new output sequence using the encoder's representations.

This is particularly useful for tasks where the input and output are different sequences, such as translation and summarization.

### Understanding cross-attention

Cross-attention is the most important architectural difference.

In decoder self-attention, queries, keys and values all come from the decoder's hidden states.

In encoder–decoder cross-attention, they come from two different sources:

**Q (Query)**: Decoder hidden states.

**K (Key)**: Encoder output.

**V (Value)**: Encoder output.

```python

from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

model_name = "facebook/bart-large-cnn"

tokenizer = AutoTokenizer.from_pretrained(model_name)
model = AutoModelForSeq2SeqLM.from_pretrained(model_name)

article = """
NASA launched a new satellite to monitor climate change.
The satellite will collect atmospheric measurements
and help scientists study global weather patterns.
"""

inputs = tokenizer(
    article,
    return_tensors="pt",
    truncation=True
)

summary_ids = model.generate(
    **inputs,
    max_new_tokens=50,
    num_beams=4
)

summary = tokenizer.decode(
    summary_ids[0],
    skip_special_tokens=True
)

print(summary)
```
